Practice 3

 In C or Java, why would `int x = 3; x = 4.5;` fail to compile, but the equivalent code runs fine in Python? Answer in 2-3 sentences.

In C and Java, variables have a fixed data type, so an int variable can only store integer values. Assigning 4.5 (a floating-point value) to an int causes a type mismatch, resulting in a compile-time error. In Python, variables are dynamically typed, so the same variable can hold values of different data types, allowing x to change from an integer to a float without any error.
