```python
import pandas as pd
data={
    'Name':['Jai', 'Princi', 'Gaurav', 'Anuj'],
    'Age':[27, 24, 22, 32],
    'City':['Delhi', 'Kanpur', 'Allahabad', 'Kannauj'],
}
df=pd.DataFrame(data)
print(df)
```

```python
student_df=pd.DataFrame(data)
print(student_df)
```

```python
city_series = pd.Series([student_df['City'][0], student_df['City'][3]], index=[0, 3])
print(city_series)
```
